***mengecek apakah dataset gabungan sudah memiliki kolom umkm/non***

```python
import pandas as pd

# pastikan df sudah punya kolom "umkm_label" berisi "UMKM" dan "NON_UMKM"
df = pd.read_csv("../olahData/nondup_labeled_dataset.csv")

n_each = 15
seed = 42  # boleh diganti biar randomnya beda

umkm_sample = (
    df[df["umkm_label"] == "UMKM"]
    .sample(n=n_each, random_state=seed)
)

non_umkm_sample = (
    df[df["umkm_label"] == "NON_UMKM"]
    .sample(n=n_each, random_state=seed)
)

sample_30 = pd.concat([umkm_sample, non_umkm_sample], ignore_index=True)

# opsional: acak lagi urutannya biar campur
sample_30 = sample_30.sample(frac=1, random_state=seed).reset_index(drop=True)

# tampilkan kolom yang biasanya penting untuk cek manual
cols = ["name","shop_name","label_titles","shop_city","shop_tier","countSold","umkm_score","umkm_reasons","umkm_label"]
print(sample_30[cols])

# kalau mau simpan buat anotasi manual:
# sample_30[cols].to_csv("sample_30_umkm_vs_non_umkm.csv", index=False)
```
