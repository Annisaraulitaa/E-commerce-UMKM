## Mengecek dan Menghapus Duplikat
**Input:** DataFrame `df` dengan kolom `id`.

**Output:** Jumlah baris duplikat berdasarkan kolom `id` dan DataFrame `df` tanpa baris duplikat.

```python
# =========================
# Hitung jumlah duplikat berdasarkan 'id'
# =========================
duplicated_rows_count = df.duplicated(subset=["id"]).sum()
duplicated_id_count = df.loc[df.duplicated(subset=["id"], keep=False), "id"].nunique()
total_rows_in_duplicate_groups = df.duplicated(subset=["id"], keep=False).sum()

print(f"Jumlah baris duplikat berdasarkan 'id'        : {duplicated_rows_count}")
print(f"Jumlah nilai 'id' yang terduplikasi           : {duplicated_id_count}")
print(f"Total baris yang termasuk grup duplikat 'id'  : {total_rows_in_duplicate_groups}")
```

```python
# =========================
# Tampilkan semua baris yang memiliki 'id' duplikat
# =========================
dup_rows = (
    df[df.duplicated(subset=["id"], keep=False)]
    .sort_values(by=["id"])
    .copy()
)

print("\nContoh baris duplikat berdasarkan 'id':")
print(dup_rows.head(20))
```

```python
# =========================
# Tampilkan frekuensi tiap 'id' yang duplikat
# =========================
dup_summary = (
    df["id"]
    .value_counts()
    .reset_index()
)

dup_summary.columns = ["id", "jumlah_kemunculan"]
dup_summary = dup_summary[dup_summary["jumlah_kemunculan"] > 1]

print("\nRingkasan 'id' yang duplikat:")
print(dup_summary.head(20))
```

```python
# Ambil hanya grup id yang duplikat
dup_rows = df[df.duplicated(subset=["id"], keep=False)].sort_values(["id"]).copy()

# Kelompokkan dan lihat ukuran tiap grup
dup_group_size = dup_rows.groupby("id").size().reset_index(name="jumlah_kemunculan")

print(dup_group_size.head(20))
```

```python
# simpan hasil ke CSV
dup_rows.to_csv("duplicate_rows_by_id.csv", index=False)
dup_summary.to_csv("duplicate_id_summary.csv", index=False)
```

## Membersihkan Teks
**Input:** Kolom teks seperti `name`, `category_breadcrumb`, `shop_city`, dan `labelGroups`.

**Output:** Kolom baru seperti `name_clean`, `category_clean`, `city_clean`, dan `labelGroups_clean` yang sudah dibersihkan dari karakter tidak diinginkan.

```python
df = pd.read_csv("../olahData/nondup_dataset.csv")
print(df.shape)
df.head()
```

```python
df.info()
```

```python
df = df.drop(columns=["mediaURL_image_fixed", "isTopAds", "labelGroups", "image_source", "category", "completeness_score"])
```

```python
# ==========================
# Untuk membersihkan teks
# ==========================
def clean_text(text):
    if pd.isna(text):
        return ""
    text = text.lower()                             # Jadikan semua huruf kecil
    text = re.sub(r"[^a-z0-9\s]", " ", text)        # Ganti semua karakter selain huruf a-z, angka 0-9, dan spasi menjadi spasi
    text = re.sub(r"\s+", " ", text).strip()        # Rapikan spasi berlebihan jadi satu spasi, lalu hapus spasi di awal/akhir
    return text
```

```python
# Bersihkan kolom ....
df["name_clean"] = df["name"].apply(clean_text)
df["category_clean"] = df["category_breadcrumb"].apply(clean_text)
df["city_clean"] = df["shop_city"].apply(clean_text)
```

```python
df[["name_clean", "category_clean", "city_clean",]].head()
```

## Normalisasi 
**Input:** Kolom `price_original` yang berisi data harga dalam format string.

**Output:** Kolom `price_original` yang sudah dinormalisasi menjadi tipe data numerik dan kolom `discountPercentage` yang menunjukkan persentase diskon.

```python
# =========================
# normalisasi kolom price_original
# =========================
df["price_original"] = (
    df["price_original"]
        .astype("string")
        .str.replace(r"[^0-9]", "", regex=True)
        .fillna("0")
        .replace("", "0")
        .astype("Int64")
)
```

```python
# Hitung discountPercentage
df["discountPercentage"] = np.where(
    df["price_original"].notna() & (df["price_original"] > 0),
    ((df["price_original"] - df["price_number"]) / df["price_original"] * 100),
    0                   # ← kalau price_original NaN / 0
)

# Rapikan hasil
df["discountPercentage"] = (
    df["discountPercentage"]
        .round(0)        # bisa ganti round(2)
        .clip(0, 100)    # jaga tetap masuk akal
        .astype(int)     # opsional: jadi integer
)

# cek hasil
df[["price_number", "price_original", "discountPercentage"]].head(10)
```

```python
# ubah ke int
cols = ["countSold", "countReview", "totalRating"]

df[cols] = df[cols].fillna(0).astype("Int64")
```

## Menghitung Skor UMKM
**Input:** Kolom seperti `countSold`, `isTopAds`, `shop_tier`, `label_titles`, `name_clean`, dan `shop_name`.

**Output:** Kolom `umkm_score` yang menunjukkan skor UMKM berdasarkan kriteria tertentu dan kolom `is_umkm` yang menunjukkan apakah produk termasuk UMKM atau tidak.

```python
UMKM_KEYWORDS = [
    # identitas & lokal
    "umkm","lokal","daerah","khas","asli","tradisional","nusantara",
    "makassar","toraja","bugis","bali","jogja","solo","bandung","malang","aceh",  # opsional: tambah daerah relevan

    # handmade & craft
    "handmade","hand made","buatan tangan","kerajinan","craft","artisan","rajut","anyaman",
    "batik","tenun","songket","ukir","gerabah","keramik","rotan","bambu",

    # rumahan & produksi kecil
    "home made","homemade","rumahan","home industry","industri rumahan","produksi rumahan",

    # custom/personalized
    "custom","kustom","preorder","pre order","made by order","by order",
]

NON_KEYWORDS = [
    "official","official store","officialstore","toko resmi","store resmi",
    "mall","tokopedia mall","brand store","flagship","flagship store",
    "authorized","authorised","distributor resmi","import","garansi resmi"
]

def make_regex(words):
    pats = []
    for w in words:
        w = w.strip().lower()
        # boundary untuk kata tunggal biar ga false match
        if " " in w:
            pats.append(re.escape(w))
        else:
            pats.append(r"\b" + re.escape(w) + r"\b")
    return re.compile("(" + "|".join(pats) + ")", re.IGNORECASE)

RE_UMKM   = make_regex(UMKM_KEYWORDS)
RE_NON    = make_regex(NON_KEYWORDS)
```

```python
def umkm_score(row):
    score = 0.0
    reasons = []

    # gabung teks untuk deteksi keyword
    text = " ".join([
        str(row.get("name_clean","") or ""),
        str(row.get("shop_name","") or ""),
        str(row.get("label_titles","") or ""),
        str(row.get("category_clean","") or ""),
        str(row.get("city_clean","") or ""),
    ]).lower()

    # --- penalti kuat non-UMKM ---
    if RE_NON.search(text):
        score -= 4
        reasons.append("official/mall")

    # --- sinyal UMKM dari keyword ---
    if RE_UMKM.search(text):
        score += 2
        reasons.append("umkm_keywords")

    # --- penjualan (skala) ---
    sold = row.get("countSold", np.nan)
    if pd.notna(sold):
        try:
            sold = float(sold)
            if sold < 50_000:
                score += 2
                reasons.append("sold<50k")
            elif sold >= 200_000:
                score -= 2
                reasons.append("sold>=200k")
        except:
            pass

    # --- tier toko ---
    tier = row.get("shop_tier", np.nan)
    if pd.notna(tier):
        try:
            tier = int(tier)
            if tier <= 2:
                score += 1
                reasons.append("tier<=2")
            elif tier >= 3:
                score -= 1
                reasons.append("tier>=3")
        except:
            pass

    # --- label "beli lokal" ---
    if "beli lokal" in text:
        score += 1
        reasons.append("beli_lokal")

    return pd.Series({
        "umkm_score": score,
        "umkm_reasons": "|".join(reasons)
    })
```

```python
df[["umkm_score","umkm_reasons"]] = df.apply(umkm_score, axis=1)

# threshold (bisa kamu tuning)
df["umkm_label"] = np.where(df["umkm_score"] >= 2, "UMKM", "NON_UMKM")
df["umkm_label"].value_counts()

# df["umkm_label"] = np.where(df["umkm_score"] >= 2, "UMKM",
#                     np.where(df["umkm_score"] <= 0, "NON_UMKM", "AMBIGUOUS"))

# df["umkm_label"].value_counts()
```
