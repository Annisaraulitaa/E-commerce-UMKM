***menggabungkan file-file dataset yang telah di scrape***

```python
import pandas as pd
import glob, os
from pathlib import Path

folder = r"d:\Kuliah\data TA\output"  # ganti sesuai lokasi folder csv kamu
files = sorted(glob.glob(os.path.join(folder, "*_enriched.csv")))  # atau "*.csv"

print("Jumlah file ketemu:", len(files))
print("Contoh:", files[:3])

dfs = []
for f in files:
    df = pd.read_csv(f)

    # opsional: biar tahu asal data
    df["source_file"] = Path(f).name
    df["category"] = Path(f).stem.replace("_enriched", "")

    dfs.append(df)

gabungan = pd.concat(dfs, ignore_index=True, sort=False)
out = os.path.join(folder, "gabungan_dataset.csv")
gabungan.to_csv(out, index=False, encoding="utf-8-sig")

print("Selesai. Output:", out, "shape:", gabungan.shape)
```
