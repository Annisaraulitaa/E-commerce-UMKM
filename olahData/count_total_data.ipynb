```python
import pandas as pd
import os
from pathlib import Path
from datetime import datetime

# Path ke folder output
output_folder = r"d:\Kuliah\data TA\output"
subfolder_z = os.path.join(output_folder, "z")

# Dictionary untuk menyimpan hasil
results = {}
grand_total = 0

print("=" * 70)
print(f"COUNTING CSV ROWS - {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
print("=" * 70)
print()

# Fungsi untuk menghitung rows
def count_csv_rows(folder_path, folder_name="Main"):
    total_rows = 0
    file_count = 0
    details = []
    
    # Cari semua file CSV
    csv_files = sorted([f for f in os.listdir(folder_path) if f.endswith('_enriched.csv')])
    
    if not csv_files:
        print(f"⚠️  Tidak ada file CSV di folder {folder_name}")
        return 0, 0, []
    
    for csv_file in csv_files:
        file_path = os.path.join(folder_path, csv_file)
        try:
            df = pd.read_csv(file_path)
            row_count = len(df)
            total_rows += row_count
            file_count += 1
            details.append({
                'file': csv_file,
                'rows': row_count,
                'columns': len(df.columns)
            })
        except Exception as e:
            print(f"❌ Error reading {csv_file}: {str(e)}")
    
    return total_rows, file_count, details

# Hitung dari folder utama
print(f"📁 FOLDER UTAMA ({output_folder})")
print("-" * 70)
main_total, main_count, main_details = count_csv_rows(output_folder, "Main")

for detail in main_details:
    print(f"  • {detail['file']:<40} | Rows: {detail['rows']:>8,} | Cols: {detail['columns']:>2}")

print(f"\n  Subtotal: {main_count} files | {main_total:,} rows")
print()

# Hitung dari subfolder z (jika ada)
print(f"📁 SUBFOLDER Z ({subfolder_z})")
print("-" * 70)
if os.path.exists(subfolder_z):
    z_total, z_count, z_details = count_csv_rows(subfolder_z, "Z")
    
    for detail in z_details:
        print(f"  • {detail['file']:<40} | Rows: {detail['rows']:>8,} | Cols: {detail['columns']:>2}")
    
    print(f"\n  Subtotal: {z_count} files | {z_total:,} rows")
    print()
else:
    z_total = 0
    z_count = 0

# Grand Total
grand_total = main_total + z_total
total_files = main_count + z_count

print("=" * 70)
print("📊 SUMMARY")
print("=" * 70)
print(f"  Total Files:        {total_files}")
print(f"  Total Rows:         {grand_total:,}")
print(f"  Generated:          {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
print("=" * 70)
```

```python
import os
from datetime import datetime

# folder_path = r"d:\Kuliah\data TA\output"
folder_path = r"D:\DATA CACA\00. College\Skripsi\E-commerce-UMKM\data lama\ulang lagi 2"   # ganti sesuai foldermu
pattern_suffix = "_enriched.csv"            # ganti kalau mau ".csv" saja

print("=" * 70)
print(f"COUNTING CSV ROWS - {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
print("=" * 70)

total_rows = 0
total_files = 0
details = []

csv_files = sorted([
    f for f in os.listdir(folder_path)
    if f.lower().endswith(pattern_suffix.lower())
])

if not csv_files:
    print(f"⚠️  Tidak ada file CSV yang cocok di: {folder_path}")
else:
    for fname in csv_files:
        fpath = os.path.join(folder_path, fname)
        try:
            with open(fpath, "r", encoding="utf-8", errors="ignore") as f:
                lines = sum(1 for _ in f)            # jumlah baris termasuk header
                rows = max(lines - 1, 0)             # baris data (tanpa header)

            total_rows += rows
            total_files += 1
            details.append((fname, rows))

        except Exception as e:
            print(f"❌ Error reading {fname}: {e}")

    # tampilkan detail
    for fname, rows in details:
        print(f"  • {fname:<45} | Rows: {rows:>10,}")

print("-" * 70)
print(f"Total Files: {total_files}")
print(f"Total Rows : {total_rows:,}")
print("=" * 70)
```

```python
# RUN CELL 2 DULU SEBELUM INI
# Hitung rows only dari _with_images.csv dan jumlah file gambar di {category}_enriched_images
from datetime import datetime
pattern_with_images = "_with_images.csv"
image_exts = {".jpg", ".jpeg", ".png", ".webp", ".bmp", ".gif", ".tif", ".tiff"}

print("=" * 120)
print(f"COUNT ROWS (with_images) + IMAGE FILES PER KATEGORI - {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
print("=" * 120)
print(f"Folder: {folder_path}")
print("-" * 120)

# Kumpulkan semua kategori berdasarkan prefix sebelum underscore dari file CSV
all_files = sorted([f for f in os.listdir(folder_path) if f.lower().endswith('.csv')])
categories = set()
for fn in all_files:
    base = os.path.splitext(fn)[0]
    if '_' in base:
        prefix = base.split('_')[0]
    else:
        prefix = base
    categories.add(prefix)

# Map rows for with_images files
rows_with_images_map = {}
csv_with_images = sorted([f for f in all_files if f.lower().endswith(pattern_with_images)])
for csv_name in csv_with_images:
    csv_path = os.path.join(folder_path, csv_name)
    base = os.path.splitext(csv_name)[0]
    category = base.split('_')[0] if '_' in base else base
    try:
        with open(csv_path, 'r', encoding='utf-8', errors='ignore') as f:
            lines = sum(1 for _ in f)
        rows = max(lines - 1, 0)
    except Exception:
        rows = 0
    rows_with_images_map[category] = rows

# Tampilkan hasil: untuk setiap kategori (prefix), tunjukkan rows_with_images dan jumlah file gambar di {category}_enriched_images
print(f"{'Kategori':<30} | {'Rows With Images':>16} | {'Images':>12} | Status")
print("-" * 120)

total_rows_with_images = 0
total_images = 0
for category in sorted(categories):
    rows_w = rows_with_images_map.get(category, 0)
    image_folder = os.path.join(folder_path, f"{category}_enriched_images")
    img_count = 0
    if os.path.isdir(image_folder):
        for fname in os.listdir(image_folder):
            fpath = os.path.join(image_folder, fname)
            ext = os.path.splitext(fname)[1].lower()
            if os.path.isfile(fpath) and ext in image_exts:
                img_count += 1
        status = '✅'
    else:
        status = '❌ folder image tidak ada'

    print(f"{category:<30} | {rows_w:>16,} | {img_count:>12,} | {status}")
    total_rows_with_images += rows_w
    total_images += img_count

print("-" * 120)
print(f"Total rows with images  : {total_rows_with_images:,}")
print(f"Total images            : {total_images:,}")
print("=" * 120)
```
